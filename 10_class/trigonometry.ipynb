# ԵՌԱՆԿՅՈՒՆԱՉԱՓՈՒԹՅՈՒՆ

## Եռանկյունաչափական ֆունկցիաների սահմանումը։

Դիտարկենք միաոր շառավիղ ունեցող շրջանագիծ որի կենտրոնը համնկնում է կորդինատային համակարգի սկզբնակետի հետ։

```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.patches import Circle, Arc
from IPython.display import Audio, display
```

```python
def create_plot():
  fig, ax = plt.subplots()
  plt.grid()
  circle = Circle((0, 0), 1, color='b', fill=False, lw=2)
  ax.add_patch(circle)
  ax.set_xlim([-1.1, 1.1])
  ax.set_ylim([-1.1, 1.1])
  ax.set_aspect('equal')
  ax.axhline(0, color='black', linestyle='-', linewidth=1.0) # X-axis
  ax.axvline(0, color='black', linestyle='-', linewidth=1.0) # Y-axis
  ax.text(0 - 0.1, 0 - 0.1, 'O', fontsize=12, color='red')
  return fig, ax

def create_vector(ax,dx_orig, dy_orig,  txt = 'M'):
  head_length = 0.1
  scale_factor = 1 - head_length
  dx_scaled = dx_orig * scale_factor
  dy_scaled = dy_orig * scale_factor
  ax.arrow(0, 0, dx_scaled, dy_scaled, head_width=0.1, head_length=head_length, fc='blue', ec='blue')
  ax.text(dx_orig + 0.05, dy_orig + 0.05, txt, fontsize=12, color='red')
  return ax

def add_proection(ax, dx_orig, dy_orig, angle_name = r'$\alpha$'):
  ax.plot([dx_orig, dx_orig], [0, dy_orig], color='gray', linestyle=':', linewidth=1) # Vertical dashed line to x-axis
  ax.plot([0, dx_orig], [dy_orig, dy_orig], color='gray', linestyle=':', linewidth=1) # Horizontal dashed line to y-axis
  ax.text(dx_orig, -0.1, "X", horizontalalignment='center', color='b')
  ax.text(-0.1, dy_orig, "Y", verticalalignment='center', color='b')
  alpha_rad = np.arctan2(dy_orig, dx_orig)
  alpha_deg = np.degrees(alpha_rad)
  arc = Arc((0, 0), 0.4, 0.4, angle=0, theta1=0, theta2=alpha_deg, color='green', lw=1, linestyle='-')
  ax.add_patch(arc)
  ax.text(0.2, 0.05, angle_name, color='green', fontsize=12)
```

```python
fig, ax = create_plot()
plt.title('(0,0) ԿԵՆՏՐՈՆՈՎ ՄԻԱՈՐ ԵՐԿԱՐՈՒԹՅԱՄԲ ՇԱՌԱՎԻՂՈՎ ՇՐՋԱՆԱԳԻԾ։')
plt.show()
```

Կորդինատային համակարգի $ O(0,0)$ սկզբնակետը կարելի է միացնել շրջանագծին պատկանող ցանկացած $ M(x,y) $ կետի և ստանալ $\vec{OM}$ միաոր երկարությամբ վեկտերը։  

```python
fig, ax = create_plot()
dx_orig = 0.5
dy_orig = (1 - dx_orig**2)**(1/2)

create_vector(ax, dx_orig, dy_orig)
plt.title('$O(0,0)$ ՍԿԶԲՆԱԿԵՏՈՎ և ՇՐՋԱՆԱԳԾԻ $M(X,Y)$ ՎԵՐՋՆԱԿԵՏՈՎ ՄԻԱՈՐ ՎԵԿՏՈՐԸ։')
plt.show()
```

Կարելի է կատարել շատ պարզ ենթադրություն այն մասին որ շրջանագիծը դա միաոր վեկտորի ծայրակետի հետագիծն է (անցած ճանապարհը) պտույտի ժամանակ։

Որպես պտույտի դրական ուղղություն վերցնենք ժամ սլաքի պտտման ուղղության հակառակը։ $X$ առանցքի դրական ուղղությունից դեպի վեկտորը պտտման դրական ուղղությամբ կազմած անկյունը նշանակենք $α$:

***Հավելում***

Կորդինատային համակարգում կետի դիրքը համարվում է տրված եթե հայտնի են կորդինատները, կամ տրված է սկղբնակետի և այդ կետը միացնող վեկտորի երկարությունը և ուղղությունը այսինքն կազմած անկյունը առանցքների հետ։ (Հարթությունում բավարար է մեկ առանցքի հետ անկյունը, տարածությունում երկու առանցքների հետ կազմած անկյունը։) Այս աշխատանքի վերջում մենք ցուից կտանք որ երկու եղանակները փոխմիարժեք են։





```python
fig, ax = create_plot()
dx_orig = 0.5
dy_orig = (1 - dx_orig**2)**(1/2)
create_vector(ax, dx_orig, dy_orig)
add_proection(ax, dx_orig, dy_orig, )
plt.title('ՎԵԿՏՈՐԻ ԿԱԶՄԱԾ ԱՆԿՅՈՒՆԸ ԴՐԱԿԱՆ ՇՐՋԱՆՑՄԱՆ ՈՒՂՂՈՒԹՅԱՄԲ։')

plt.show()
```

***
***Սահմանում***

$\alpha$ անկյան սինուս կանվանենք $M$ կետի «միաոր վեկտորի ծայրակետի» $Y$ կորդինատը նշանակվում է $\sin{(\alpha})$։

***

***Սահմանում***

$\alpha$ անկյան կոսինուս կանվանենք $M$ կետի «միաոր վեկտորի ծայրակետի» $X$ կորդինատը նշանակվում է $\cos{(\alpha})$։

***

* անկյան սինուս, կոսինուս ֆունկցիաները պարբերական են։ $ \sin{(\alpha)} = \sin{(\alpha +T)} \quad $  $ \cos{(\alpha)} = \cos{(\alpha +T)}  $  որտեղ $T$-ն պարբերությունն է։ Իսկապես դժվար չէ ցույց տալ որ եթե $T = 2 \pi n, \quad n \in N$։ Այսինքն միաոր վեկտորը կատարել է  n թվով պտույտներ գալու է նույտ տեղում կանգ առնի հետևաբար վեկտորի ծայրակետը կունենա նույն կորդինատները։

*  Ակնհայտ է որ անկյան սինուսի և կոսինուսի  արժեքները ընկած են $[-1, 1]$ միջակայքում։ Սա պարզ երևում է գծագրից։ Օրինակ $cos{(\alpha)}$֊ն իր առավելագույն արժեքը կստանա $0 + 2  \pi  n, \quad  n \in N$

* $sin$ և $cos$ ֆունկցիաները որոշված են $(- \infty, \infty)$ և աննդհատ է ամբողջ $R$ ում։ Այս պնդումը ևս կարելի է մաթեմատիկորեն խիստ ապացուցել սակայն բավարար է միայն հիշել ֆունկցիաների սահմանումները և պարզ է դառնում որ ինչպիսին էլ լինի $\alpha$ անկյունը «միաոր վեկտորի դիրքը» վեկտորի ծայրակետը կունենա $x, y$ կորդինատներ։   

* $\sin^2{α} + \cos^2{α} = 1$ առնչությունը կոչվում է եռանկյունաչափական հիմնական նույնություն։ Այս առնչությունը նույնպես դժվար չէ ապացուցել։ Դիտարկենք $OMX$ ուղղանկյուն եռանկյունը: Ըստ Պյութագորասի թյորեմի ուղղանկյուն եռանկյան էջերի քառակուսիների գումարը հավասար է ներքնաձիքի քառակուսուն։ Դիտարկված եռանկյունում էջերը ըստ մեր սահմանման $\alpha$ անկյան $\sin$ և $\cos$ ֆունկցիաների արժեքներն են, իսք ներքնաձիքի երկարությունը 1 է։

* Եթե $\alpha + β = \frac{\pi}{2}$ ապա $\sin{(\alpha)} = \cos{(\beta)}, \quad \cos{(\alpha)} = \sin{(\beta)}$։ Այս պնդումը նույնպես դժվար չէ ցույց տալ։ Դպրոցական երկրաչափությունից հայտնի է որ եռանկյան անկյունների գումարը $\pi$ է $⇒$ եթե փորձենք $\alpha$ և $\beta$ անկյուններ ունեցող եռանկյուն կառուցել ինչպիսին էլ լինեն $\alpha$-ն և $\beta$-ն դրանց գումարը $\frac{\pi}{2}$ է նշանակում է երորդ անկյունը պետք է լինի $\pi -  \frac{\pi}{2} = \frac{\pi}{2}$ ուստի կստացվի ուղղանկյուն եռանկյուն։ ակնհայտ է որ $\alpha$ անկյան դիմացի էջը կլինի $\beta$ անկյան կիցը և հակառակը։

* Հարթությունը կորդինատային առանցքները բաժանում են $4$ ենթահարթությունների։ Այդ ենթահարթությունները անվանվում են առաջին, երկրորդ, երորդ, և չորորդ քառորդներ ըստ դրական պտտման ուղղության։ այս քառորդներից յուրաքանչյուրը օժդված է հատկությամբ որը չունեն  միուս ենթահարթությունները։ Որպես այդպիսի հատկություն հանդես է գալիս այդ ենթահարթություններում պատկանող ցանկացած կետի կորդինատների նշանները։ Ակնհայտ է որ ցանկացած կետ որը առաջին քառորդում է ունի դրական կորդինատներ։ Հակառակը նույնպես ճիշտ է այսինքն այն կետը որը ունի դրական կորդինատներ կարող է գտնվել միայն և միայն առաջին քառորդում։ Երկրորդ քառուրդում կետը կգտնվի եթե ունի բացասական $X$ և դրական $Y$ կորդինատ, երորդում եթե և $X$-ը և $Y$-ը բացասական են իսք չորորդում $X>0 \quad Y < 0$: Այս պարզ պնդումը բնականաբար վերաբերվում է նաև $sin$ և $cos$ ֆունկցիաներին։        


Ընդհանրապես $y = f(x)$ կախված ֆունկցիայից ունեն մի շարք հատկություններ ըստ որոնց կարելի է տարանջատել ֆունկցիաները։ Այդպիսի հատկություններից է՝ աճող է թե նվազող, պարբերական է թե ոչ, միարժեք է թե բազմարժեք, աննդհատ է թե ոչ, զույգ է թե կենտ և այլն։ Բոլոր այս հատկությունները հայտնի են դառնում ֆունկցիայի հետազոտման ժամանակ՝ այսինքն իմանալ ֆունկցիայի պահվածքը «վարքը» արգումենտի ամբողջ $R$ բազմությունում։

***
***Սահմանում***

 f(x) ֆունկցիան կոչվում է զույգ, եթե իր արժեքը չի փոխում արգումենտի նշանը փոխելիս, այսինքն ֆունկցիան զույգ է, եթե. $f(x) = f(-x)$ և կենտ, եթե. $f(-x) = -f(x)$


***

զույգ ֆունկցիաների օրինակ է՝ $y = x^2, \quad y= |x|, \quad y = \cos(x) $ կենտ ֆունկցիաներ են՝ $y = x^3, \quad y = kx, \quad y = \sin(x)$ ֆունկցիաները։

Նախ պատկերենք նշված ֆունկցիաների գրաֆիքները հետո քննարկենք $\sin$ և $\cos$ ֆունկցիաների զույգությունը և կենտությունը։  





```python
x = np.linspace(-10, 10, 500)
x_sq = x**2
x_mod = np.abs(x)

x_cub = x**3
x_line = -5*x

plt.figure(figsize=(12, 10)) # Increased figure size for better visibility

# Subplot 1: y = x^2 (Even Function)
ax1 = plt.subplot(221)
ax1.plot(x, x_sq, color='red', label='$y = x^2$')
ax1.set_title('$y = x^2$ (Զույգ ֆունկցիա)') # Armenian for Even Function
ax1.grid(True)
ax1.axhline(0, color='black', linewidth=0.8) # X-axis
ax1.axvline(0, color='black', linewidth=0.8) # Y-axis
ax1.legend()

# Subplot 2: y = |x| (Even Function)
ax2 = plt.subplot(222)
ax2.plot(x, x_mod, color='red', label='$y = |x|$')
ax2.set_title('$y = |x|$ (Զույգ ֆունկցիա)')
ax2.grid(True)
ax2.axhline(0, color='black', linewidth=0.8) # X-axis
ax2.axvline(0, color='black', linewidth=0.8) # Y-axis
ax2.legend()

# Subplot 3: y = x^3 (Odd Function)
ax3 = plt.subplot(223)
ax3.plot(x, x_cub, color='blue', label='$y = x^3$')
ax3.set_title('$y = x^3$ (Կենտ ֆունկցիա)') # Armenian for Odd Function
ax3.grid(True)
ax3.axhline(0, color='black', linewidth=0.8) # X-axis
ax3.axvline(0, color='black', linewidth=0.8) # Y-axis
ax3.legend()

# Subplot 4: y = -5x (Odd Function)
ax4 = plt.subplot(224) # Changed from 223 to 224 to avoid overwriting
ax4.plot(x, x_line, color='blue', label='$y = -5x$')
ax4.set_title('$y = -5x$ (Կենտ ֆունկցիա)')
ax4.grid(True)
ax4.axhline(0, color='black', linewidth=0.8) # X-axis
ax4.axvline(0, color='black', linewidth=0.8) # Y-axis
ax4.legend()

plt.tight_layout() # Adjust layout to prevent overlap
plt.show()
```

Ինչպես երևում է նկարից զույգ ֆունկցիաները նրանք են որոնց գրաֆիքների համար $Y$֊ների առանցքը կարծես լինի «հայելի»։ Եթե $x>0$ բազմության գրաֆիքը պտտենք $Y$ առանցքի շուրջ կստանանք $x<0$֊ի գրաֆիքը՝ այսինքն **զույգ ֆունկցիայի գրաֆիքը սիմետրիկ է $Y$ առանցքի նկատմամբ**։ Նույնը կարելի է ասել կենտ ֆունկցիայի և $X$ առանցքի նկատմամբ։

$ y= \cos(\alpha)$ զույգությունը բխում է այն հանգամանքից որ, որպես $\alpha$ անկյուն վերցրել ենք հենց $X$֊երի առանցքի դրական մասի և ժամ սլաքի ուղղության հակառակ կազմած անկյունը։ Այսպիսով $\alpha$ և $-\alpha$ անկյուն ունեցող վեկտորները սիմետրիկ արտապատկերումներն են $X$ առանցքի նկատմամբ $⇒$ դրանց ծայրակետի $x$ կորդինատները նույնն են իսք $y$ կերդինատները տարբերվում են նշանով $⇒$ $ y= \sin(\alpha)$ կենտ է։

```python
dx_orig = 0.5 # Chosen a positive x-coordinate for better visibility of alpha and -alpha
dy_orig = (1 - dx_orig**2)**(1/2)

fig, ax = create_plot()

# Vector for alpha
create_vector(ax, dx_orig, dy_orig, txt=r'$M_\alpha$')
add_proection(ax, dx_orig, dy_orig, angle_name=r'$+\alpha$')

# Vector for -alpha
create_vector(ax, dx_orig, -dy_orig, txt=r'$M_{-\alpha}$', )

# Add arc for -alpha
alpha_rad_neg = np.arctan2(-dy_orig, dx_orig)
alpha_deg_neg = np.degrees(alpha_rad_neg)
arc_neg_alpha = Arc((0, 0), 0.4, 0.4, angle=0, theta1=alpha_deg_neg, theta2=0, color='purple', lw=1, linestyle='-')
ax.add_patch(arc_neg_alpha)
ax.text(0.2, -0.05, r'$-\alpha$', color='purple', fontsize=12)

# Add projections for -alpha
ax.plot([dx_orig, dx_orig], [0, -dy_orig], color='gray', linestyle=':', linewidth=1) # Vertical dashed line to x-axis for -alpha
ax.plot([0, dx_orig], [-dy_orig, -dy_orig], color='gray', linestyle=':', linewidth=1) # Horizontal dashed line to y-axis for -alpha

# Label X_alpha and X_{-alpha} (they are the same)
ax.text(dx_orig, -0.15, r'$x$', horizontalalignment='center', color='b')
# Label Y_alpha and Y_{-alpha}
ax.text(-0.1, dy_orig, r'$y_{\alpha}$', verticalalignment='center', color='b')
ax.text(-0.1, -dy_orig, r'$y_{-\alpha}$', verticalalignment='center', color='b')


plt.title(r'ԱՆԿՅՈՒՆՆԵՐԻ $\alpha$ ԵՎ $ -\alpha $ $\sin$-Ի ԵՎ $\cos$-Ի ԱՐԺԵՔՆԵՐԸ')
plt.show()
```

Նկարում պատկերված է երկու միաոր վեկտորները որոնք կազմում են $α$ և $֊α$ անկյուններ։ Այն որ $\cos{(\alpha)} = \cos{(- \alpha)},  \quad  \sin{(-\alpha)} =  - \sin{(\alpha)}  \quad \forall \alpha$֊ի համար երևում է վերևի նկարից։  

## ՄԻ ՔԱՆԻ ԱՆԿՅՈՒՆՆԵՐԻ $\sin$ և $\cos$ ԱՐԺԵՔՆԵՐԸ։

այժմ տանք մի քանի անկյունների $\sin$ և $\cos$ արժեքները։

երբ $\alpha = 0$ միաոր վեկտորը ընկած է $X$ առանցքի երկայնքով $\Rightarrow$ $\sin{(0)} = 0$, իսք $\cos(0) = 1$: Նույն կերպ երբ $\alpha = \frac{\pi}{2}$ վեկտորը ընկած կլինի $Y$ առանցքի երկայնքով $\Rightarrow$ $\cos{(\frac{\pi}{2})} = 0$, իսք $\sin{(\frac{\pi}{2})} = 1$ $sin{(\pi)} = 0 , \quad cos{(\pi)} = -1, \quad sin{(\frac{3\pi}{2})} = -1, \quad cos{(\frac{3\pi}{2})} = 0 $ ։

```python
dy_orig = 0.5
dx_orig = (1 - dy_orig**2)**(1/2)
fig, ax = create_plot()
create_vector(ax, dx_orig, dy_orig)
add_proection(ax, dx_orig, dy_orig, angle_name=r'$\alpha = \frac{\pi}{6}$' )

ax.plot([dx_orig, dx_orig, 0], [dy_orig, dy_orig-1, 0], color = 'b', linewidth = 1)

ax.text(dx_orig + 0.05, dy_orig-1.05, 'N', fontsize=12, color='red')

plt.title(r"$\frac{\pi}{6}$ ԱՆԿՅԱՆ $\sin$-Ի և $\cos$-Ի ՀԱՇՎԱՐԿԸ։ ")
plt.show()
```

Որպիսի գտնենք $\frac{\pi}{6}$ անկյան սինուսի և կոսինուսի արժեքները դիտարկենք նկարում պատկերված $OMN$ եռանկյունը։ $OM = ON = 1$ շրջանագծի շառավիղն է  $\angle OMN = \frac{pi}{3}$ քանի որ $OMX$ ը ուղղանկյուն եռանկյուն է։ $⇒$ $OMN$-ը հավասարակողմ եռանկյուն է։ իսք $Օx$-ը  հանդիսանում է $NM$ ի միջնուղղահայաց։ նշվածից հետևում է որ $Mx = xN = \sin{(\frac{\pi}{6})} = \frac{1}{2}$: (Այս եզրահանգումը ընթերցողին կարող է ծանոթ լինել դպրոցական երկրաչափությունից նաև այլ կերպ։ ՈՒղղանկյուն եռանկյան $\frac{\pi}{6}$ անկյան դիմաց ընկած էջը հավասար է ներքնաձիքի կեսին։) Օգտվելով եռանկյունաչափական հիմնական նույնությունից կարող ենք հաշվել նաև $cos{(\frac{\pi}{6})} = \sqrt{1 - \frac{1}{4}} = \frac{\sqrt{3}}{2}$: Ընթերցողին առաջարկում ենք հաշվել  $\frac{5 \pi}{6}, \frac{7 \pi}{6}, \frac{11 \pi}{6}$ անկյունների $\sin$ և $\cos$ արժեքները։

Այժմ գտնենք $\frac{\pi}{4}$ անկյան $\sin$-ը և $\cos$-ը։ Սկզբից պատկերենք միաոր վեկտոր որը $X$-էրի հետ կազմում է $\frac{\pi}{4}$ անկյուն։

```python
dx_orig = (2**(1/2))/2
dy_orig = (2**(1/2))/2
fig, ax = create_plot()
create_vector(ax, dx_orig, dy_orig)
add_proection(ax, dx_orig, dy_orig, angle_name=r'$\alpha = \frac{\pi}{4}$' )

ax.plot([dx_orig, dx_orig], [0, dy_orig], color='b', linestyle='-', linewidth=1) # Vertical dashed line to x-axis
ax.plot([0, dx_orig], [dy_orig, dy_orig], color='b', linestyle='-', linewidth=1) # Horizontal dashed line to y-axis

plt.title(r"$\frac{\pi}{4}$ ԱՆԿՅԱՆ $\sin$-Ի և $\cos$-Ի ՀԱՇՎԱՐԿԸ։ ")
plt.show()
```

$OYMX$ քառանկյան բոլոր անկյունները $\frac{\pi}{2}$ է, իսք անկյունագիծը կողմերի հետ կազմում է $\frac{\pi}{4}$ անկյուն $⇒$ այն քառակուսի է։ Նշանակուն է $x = y$ այսինքն $\sin{(\frac{\pi}{4})} = \cos{(\frac{\pi}{4})} $: Օգտվելով հիմնական նույնությունից կստանանք $\sin{(\frac{\pi}{4})} = \cos{(\frac{\pi}{4})} = \frac{\sqrt{2}}{2}$: Դժվար չէ գտնել $\frac{\pi}{4} + n \frac{\pi}{2}, \quad n \in N $ անկյունների $\sin$-ը և $\cos$-ը։

$\frac{\pi}{3} $ անկյան $\sin$ և $\cos$ ֆունկցիաների արժեքների հաշվարկը կթողնենք ընթերցողին։

Այսպիսով ստացված արդյունքները կարող ենք գրել աղյուսակի տեսքով՝

## $\sin$ և $\cos$ ՖՈՒՆԿՑԻԱՆԵՐԻ ԳՐԱՖԻԿՆԵՐԸ։ ՁԱՅՆԸ ՈՐՊԵՍ ՆԵՐԴԱՇՆԱԿ ՖՈՒՆԿՑԻԱ։

Այժմ ստանանք դիտարկված ֆունկցիաների գրաֆիկը։ Որպես անկյան արժեքներ վերցնենք $[-2\pi, 2\pi]$ միջակայքը։


```python
alpha = np.linspace(-2*np.pi, 2*np.pi, 720)
sin_alp = np.sin(alpha)

# Redefine famous_alpha to include all standard angles and ensure it's sorted
first_quadrant_famus = np.array([0, np.pi/6, np.pi/4, np.pi/3, np.pi/2])

# Generate all positive famous angles up to 2*pi
all_positive_famous = np.array([])
for q_mult in range(4):
    all_positive_famous = np.concatenate((all_positive_famous, first_quadrant_famus + q_mult * np.pi/2))

# Add 2*pi explicitly if not already present and ensure uniqueness and sorting
all_positive_famous = np.unique(all_positive_famous)
all_positive_famous = all_positive_famous[all_positive_famous <= 2*np.pi] # Filter out anything slightly above 2pi due to precision
if not np.isclose(all_positive_famous[-1], 2*np.pi): # Add 2*pi if not already close enough
    all_positive_famous = np.append(all_positive_famous, 2*np.pi)

# Combine positive and negative famous angles and sort
famous_alpha = np.concatenate((-all_positive_famous, all_positive_famous))
famous_alpha = np.unique(famous_alpha)
famous_alpha.sort()

sin_famous = np.sin(famous_alpha)



plt.figure(figsize=(12, 8)) # Increased figure size
plt.subplot(211)
plt.plot(famous_alpha, sin_famous, color='tab:blue', marker='o', linestyle='None') # Changed linestyle to None
plt.plot(alpha, sin_alp, color='black',)
plt.title('$\\sin(\\alpha)$ function') # Title for the first subplot
plt.grid(True) # Add grid

plt.subplot(212)
plt.plot(famous_alpha, np.cos(famous_alpha), color='tab:blue', marker='o', linestyle='None') # Changed linestyle to None
plt.plot(alpha, np.cos(alpha), color='black',)
plt.title('$\\cos(\\alpha)$ function') # Title for the second subplot
plt.grid(True) # Add grid
plt.show()
```

Որպես գործնական օրինակ դիտարկենք ձայնը որի հետ մենք բոլորս ամեն օր առնչվում ենք։ Ինչպես գիտենք ձայնը օդի տատանում է որը կարող է առաջանալ տարբեր եղանակներով (մարդու ձայնալարերի տատանումները, բարձրախոսի մեմբրանի տատանումները, ․․․)։ Մարդու ականջները կարող են լսել միջինում 20 ից 20․000 հց հաճախությամբ տատանումներ «կան նաև բացառություններ»։ Մի փոքր այն մասին թե ինչ է  հաճախականությունը՝ դա միաոր ժամանակում կատարած լրիվ տատանումների թիվն է։ Մեկ տատանումը դա այն միջակայքն է երբ մեծությունը սկսում է կրկնել իր նախքինում ունեցած արժեքները։ Այժմ դիտարկենք $y = A_0\sin{(ωt)}$, $t$ ժամանակից կախված ֆունկցիան։ $\omega$֊ն կոչվում է անկյունային հաճախություն ակնհայտ է որ որքան մեծացնենք $\omega$֊ն այդքան շատ կլինի միաոր ժամանակում կատարած տատանումների թիվը, $A_0$֊ն կոչվում է ամպլիտուդ այստեղ էլ ակնհայտ է որ $y \in [-A_0, A_0 ]$։ Ասվածի պարզեցման համար պատկերենք երկու ուսանելի տատանումներ որից առաջինը ժամանակի ընթացքում արագանում է։ Իսք երկրորդը մարում է։ Այլ կերպ ասած առաջինում $\omega$ ֊ն աճող ֆունկցիա է ժամանակից $\omega(t) = \omega_0 + kt \quad k>0$ Իսք երկրորդի ամպլիտուդն է նվազող ֆունկցիա ժամանակից։      

```python
# Parameters for the plots
sampling_rate = 1000 # samples per second
duration = 5       # seconds
t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)

# --- First plot: Increasing frequency (accelerating oscillation) ---
omega_0_1 = 2 * np.pi * 1  # Initial angular frequency (1 Hz)
k_freq = 2 * np.pi * 0.5   # Rate of angular frequency increase (0.5 Hz/s)
amplitude_1 = 1.0

# instantaneous angular frequency: omega(t) = omega_0 + k*t
# integral of omega(t) gives the phase: phi(t) = omega_0*t + 0.5*k*t^2
# For simplicity, let's directly use a time-varying frequency in the sine argument
# This is an approximation for rapidly changing frequencies but good for illustration
accelerating_oscillation = amplitude_1 * np.sin(omega_0_1 * t + 0.5 * k_freq * t**2)

# --- Second plot: Damping oscillation (decreasing amplitude) ---
frequency_2 = 2.0  # Constant frequency (2 Hz)
amplitude_0_2 = 1.0 # Initial amplitude for damping
k_amp = 0.5         # Decay rate (positive value for damping)

amplitude_over_time = amplitude_0_2 * np.exp(-k_amp * t)

damping_oscillation = amplitude_over_time * np.sin(2 * np.pi * frequency_2 * t)

# Plotting
plt.figure(figsize=(12, 10))

plt.subplot(2, 1, 1)
plt.plot(t, accelerating_oscillation, color='blue')
plt.title('Increasing Frequency Oscillation ($\omega(t) = \omega_0 + kt$)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(t, damping_oscillation, color='red') # Use damping_oscillation here
plt.title('Damping Oscillation ($A(t) = A_0 e^{-kt}$)') # Updated title
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

Չշեղվելով թեմայից ասենք որ երաժշտական նոտաները դրանք սինուսոիդներ (ժամանակից կախված սինուս ֆունկցիա) էն տարբեր հիմնական հաճախություններով և հարմոնիկներից ( հարմոնիկների մասին կարող եք գտնել տեղեկություն համացանցից )։ այժմ ստանանք հինգերորդ օկտավաի դո նոտան պատկերենք այն 10 մվ ժամանակով հետո լսենք։

```python
# Configuration for C5 note (5th octave)
frequency_hz = 523.25  # Frequency for C5 in Hz
sampling_rate = 44100 # standard CD quality samples per second
duration = 2.0        # seconds

# Generate the time array for audio samples
t_audio = np.linspace(0., duration, int(sampling_rate * duration), endpoint=False)

# Generate the sine wave samples (amplitude from -1.0 to 1.0)
# Formula: amplitude * sin(2 * pi * frequency * time)
samples =  np.sin(2. * np.pi * frequency_hz * t_audio)

# Plot the generated sine wave - now focusing on a very short segment
plt.figure(figsize=(10, 4))

# Define a short duration for plotting (e.g., 10 milliseconds)
plot_duration_ms = 10
plot_samples = int(sampling_rate * (plot_duration_ms / 1000.0))

# Ensure we don't try to plot more samples than available
plot_samples = min(plot_samples, len(samples))

t_plot = np.linspace(0., plot_duration_ms / 1000.0, plot_samples, endpoint=False)

plt.plot(t_plot, samples[:plot_samples], color='blue')
plt.title(f'Waveform of C5 Note ({frequency_hz} Hz) - First {plot_duration_ms} ms')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()

print(f"Generating C5 note at {frequency_hz} Hz for {duration} seconds...")
display(Audio(samples, rate=sampling_rate))
print("Playback initiated.")
```

### ԵՐԿՈՒ ԱՆԿՅՈՒՆՆԵՐԻ ԳՈՒՄԱՐԻ $\sin$-Ը և $\cos$-Ը։

Այժմ դիտարկենք որևէ $\gamma \neq 0 $ անկյուն որը ներկայացնենք երկու $\alpha$ և $\beta$  անկյունների գումար։ Նախ պատկերենք միաոր շրջանագիծ և երկու կորդինատային համակարգ որոնց սկզբնակետը համնկնում է շրջանագծի կենտրոնի հետ իսք իրար նկատմամբ պտտված են $\alpha$ անկյամբ։

```python
fig, ax = create_plot()

alpha_rad = np.pi / 6
dx_alpha = np.cos(alpha_rad)
dy_alpha = np.sin(alpha_rad)

ax.plot([-1.1, 1.1], [-1.1*dy_alpha/dx_alpha,1.1*dy_alpha/dx_alpha], c = 'olive')
ax.plot([1.1*dy_alpha/dx_alpha, -1.1*dy_alpha/dx_alpha], [-1.1, 1.1], c = 'olive')

# Add labels for X and Y axes
ax.text(1.05, -0.08, 'X', color='black', fontsize=12)
ax.text(-0.05, 1.05, 'Y', color='black', fontsize=12)

# Add labels for rotated X' and Y' axes
ax.text(1.1 * dx_alpha, 0.9 * dy_alpha, r"X'", color='olive', fontsize=12)
ax.text(-1.05 * dy_alpha, 1.05 * dx_alpha, r"Y'", color='olive', fontsize=12)

alpha_deg = np.degrees(alpha_rad)
arc_alpha = Arc((0, 0), 0.4, 0.4, angle=0, theta1=0, theta2=alpha_deg, color='green', lw=1, linestyle='-')
ax.add_patch(arc_alpha)
ax.text(0.2, 0.05, r'$\alpha$', color='green', fontsize=12)

plt.title(r"ԵՐԿՈՒ ԿՈՐԴԻՆԱՏԱՅԻՆ ՀԱՄԱԿԱՐԳԵՐ ՊՏՏՎԱԾ $\alpha$ ԱՆԿՅԱՄԲ")
plt.show()
```
