# ԵՌԱՆԿՅՈՒՆԱՉԱՓՈՒԹՅՈՒՆ

## Եռանկյունաչափական ֆունկցիաների սահմանումը։

Դիտարկենք միաոր շառավիղ ունեցող շրջանագիծ որի կենտրոնը համնկնում է կորդինատային համակարգի սկզբնակետի հետ։

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Arc
from IPython.display import Audio, display
```

```python
def create_plot():
  fig, ax = plt.subplots()
  fig.set_size_inches(10, 8) # Correctly resize the figure
  plt.grid()
  circle = Circle((0, 0), 1, color='b', fill=False, lw=2)
  ax.add_patch(circle)
  ax.set_xlim([-1.1, 1.1])
  ax.set_ylim([-1.1, 1.1])
  ax.set_aspect('equal')
  ax.axhline(0, color='black', linestyle='-', linewidth=1.0) # X-axis
  ax.axvline(0, color='black', linestyle='-', linewidth=1.0) # Y-axis
  ax.plot(0, 0, 'o', color='black', markersize=5)
  ax.text(0 - 0.1, 0 - 0.1, 'O', fontsize=12, color='black')
  return fig, ax

def create_vector(ax,dx_orig, dy_orig, x_deviation = 0.05, y_deviation = 0.05,  txt = 'M'):
  head_length = 0.1
  scale_factor = 1 - head_length
  dx_scaled = dx_orig * scale_factor
  dy_scaled = dy_orig * scale_factor
  ax.arrow(0, 0, dx_scaled, dy_scaled, head_width=0.1, head_length=head_length, fc='blue', ec='blue')
  ax.plot(dx_orig, dy_orig, 'o', color='blue', markersize=5)
  ax.text(dx_orig + x_deviation, dy_orig + y_deviation, txt, fontsize=12, color='blue')
  return ax

def add_proection(ax, dx_orig, dy_orig, x_name = "X", y_name = "Y", size_angke = 0.4, angle_name_x=0.2, angle_name_y=0.05,  angle_name = r'$\alpha$', colr = "green", pls = True,):
  ax.plot([dx_orig, dx_orig], [0, dy_orig], color='gray', linestyle=':', linewidth=1.5) # Vertical dashed line to x-axis
  ax.plot([0, dx_orig], [dy_orig, dy_orig], color='gray', linestyle=':', linewidth=1.5) # Horizontal dashed line to y-axis
  ax.plot(0, dy_orig, 'o', color='black', markersize=5)
  ax.plot(dx_orig, 0, 'o', color='black', markersize=5)
  ax.text(dx_orig, -0.1, x_name, horizontalalignment='center', color='b')
  ax.text(-0.1, dy_orig, y_name, verticalalignment='center', color='b')

  alpha_rad = np.arctan2(dy_orig, dx_orig)
  alpha_deg = np.degrees(alpha_rad)
  if pls:
    arc = Arc((0, 0), size_angke, size_angke, angle=0, theta1=0, theta2=alpha_deg, color=colr, lw=1, linestyle='-')
  else:
    arc = Arc((0, 0), size_angke, size_angke, angle=0, theta1=alpha_deg, theta2=0, color=colr, lw=1, linestyle='-')

  ax.add_patch(arc)
  ax.text(angle_name_x, angle_name_y, angle_name, color=colr, fontsize=12)
```

```python
fig, ax = create_plot()
plt.title('(0,0) ԿԵՆՏՐՈՆՈՎ ՄԻԱՈՐ ԵՐԿԱՐՈՒԹՅԱՄԲ ՇԱՌԱՎԻՂՈՎ ՇՐՋԱՆԱԳԻԾ։')
plt.show()
```

Կորդինատային համակարգի $ O(0,0)$ սկզբնակետը կարելի է միացնել շրջանագծին պատկանող ցանկացած $ M(x,y) $ կետի և ստանալ $\vec{OM}$ միաոր երկարությամբ վեկտերը։  

```python
fig, ax = create_plot()
dx_orig = 0.5
dy_orig = (1 - dx_orig**2)**(1/2)

create_vector(ax, dx_orig, dy_orig)
plt.title('$O(0,0)$ ՍԿԶԲՆԱԿԵՏՈՎ և ՇՐՋԱՆԱԳԾԻ $M(X,Y)$ ՎԵՐՋՆԱԿԵՏՈՎ ՄԻԱՈՐ ՎԵԿՏՈՐԸ։')
plt.show()
```

Կարելի է կատարել շատ պարզ ենթադրություն այն մասին որ շրջանագիծը դա միաոր վեկտորի ծայրակետի հետագիծն է (անցած ճանապարհը) պտույտի ժամանակ։

Որպես պտույտի դրական ուղղություն վերցնենք ժամ սլաքի պտտման ուղղության հակառակը։ $X$ առանցքի դրական ուղղությունից դեպի վեկտորը պտտման դրական ուղղությամբ կազմած անկյունը նշանակենք $α$:

***Հավելում***

Կորդինատային համակարգում կետի դիրքը համարվում է տրված եթե հայտնի են կորդինատները, կամ տրված է սկղբնակետի և այդ կետը միացնող վեկտորի երկարությունը և ուղղությունը այսինքն կազմած անկյունը առանցքների հետ։ (Հարթությունում բավարար է մեկ առանցքի հետ անկյունը, տարածությունում երկու առանցքների հետ կազմած անկյունը։) Այս աշխատանքի վերջում մենք ցուից կտանք որ երկու եղանակները փոխմիարժեք են։



```python
fig, ax = create_plot()
dx_orig = 0.5
dy_orig = (1 - dx_orig**2)**(1/2)
create_vector(ax, dx_orig, dy_orig)
add_proection(ax, dx_orig, dy_orig, )
plt.title('ՎԵԿՏՈՐԻ ԿԱԶՄԱԾ ԱՆԿՅՈՒՆԸ ԴՐԱԿԱՆ ՇՐՋԱՆՑՄԱՆ ՈՒՂՂՈՒԹՅԱՄԲ։')

plt.show()
```

***
***Սահմանում***

$\alpha$ անկյան սինուս կանվանենք միաոր վեկտորի ծայրակետի «$M$ կետի» $Y$ կորդինատը նշանակվում է $\sin{(\alpha})$։

***

***Սահմանում***

$\alpha$ անկյան կոսինուս կանվանենք միաոր վեկտորի ծայրակետի  «$M$ կետի» $X$ կորդինատը նշանակվում է $\cos{(\alpha})$։

***

* անկյան սինուս, կոսինուս ֆունկցիաները պարբերական են։ $ \sin{(\alpha)} = \sin{(\alpha +T)} \quad $  $ \cos{(\alpha)} = \cos{(\alpha +T)}  $  որտեղ $T$-ն պարբերությունն է։ Իսկապես դժվար չէ ցույց տալ որ եթե $T = 2 \pi k, \quad k \in Z$։ Այսինքն միաոր վեկտորը կատարելով k թվով պտույտներ գալու է նույտ տեղում կանգ առնի հետևաբար վեկտորի ծայրակետը կունենա նույն կորդինատները։

*  Ակնհայտ է որ անկյան սինուսի և կոսինուսի  արժեքները ընկած են $[-1, 1]$ միջակայքում։ Սա պարզ երևում է գծագրից։ Օրինակ $cos{(\alpha)}$֊ն իր առավելագույն արժեքը կստանա $0 + 2  \pi  k, \quad  k \in Z$

* $sin$ և $cos$ ֆունկցիաները որոշված են $(- \infty, \infty)$ և աննդհատ է ամբողջ $R$ ում։ Այս պնդումը ևս կարելի է մաթեմատիկորեն խիստ ապացուցել սակայն բավարար է միայն հիշել ֆունկցիաների սահմանումները և պարզ է դառնում որ ինչպիսին էլ լինի $\alpha$ անկյունը «միաոր վեկտորի դիրքը» վեկտորի ծայրակետը կունենա $x, y$ կորդինատներ։   

* $\sin^2{α} + \cos^2{α} = 1$ առնչությունը կոչվում է եռանկյունաչափական հիմնական նույնություն։ Այս առնչությունը նույնպես դժվար չէ ապացուցել։ Դիտարկենք $OMX$ ուղղանկյուն եռանկյունը: Ըստ Պյութագորասի թյորեմի ուղղանկյուն եռանկյան էջերի քառակուսիների գումարը հավասար է ներքնաձիքի քառակուսուն։ Դիտարկված եռանկյունում էջերը ըստ մեր սահմանման $\alpha$ անկյան $\sin$ և $\cos$ ֆունկցիաների արժեքներն են, իսք ներքնաձիքի երկարությունը 1 է։

* Եթե $\alpha + β = \frac{\pi}{2}$ ապա $\sin{(\alpha)} = \cos{(\beta)}, \quad \cos{(\alpha)} = \sin{(\beta)}$։ Այս պնդումը նույնպես դժվար չէ ցույց տալ։ Դպրոցական երկրաչափությունից հայտնի է որ եռանկյան անկյունների գումարը $\pi$ է $⇒$ եթե փորձենք $\alpha$ և $\beta$ անկյուններ ունեցող եռանկյուն կառուցել ինչպիսին էլ լինեն $\alpha$-ն և $\beta$-ն դրանց գումարը $\frac{\pi}{2}$ է նշանակում է երորդ անկյունը պետք է լինի $\pi -  \frac{\pi}{2} = \frac{\pi}{2}$ ուստի կստացվի ուղղանկյուն եռանկյուն։ ակնհայտ է որ $\alpha$ անկյան դիմացի էջը կլինի $\beta$ անկյան կիցը և հակառակը։

* Հարթությունը կորդինատային առանցքները բաժանում են $4$ ենթահարթությունների։ Այդ ենթահարթությունները անվանվում են առաջին, երկրորդ, երորդ, և չորորդ քառորդներ ըստ դրական պտտման ուղղության։ այս քառորդներից յուրաքանչյուրը օժդված է հատկությամբ որը չունեն  միուս ենթահարթությունները։ Որպես այդպիսի հատկություն հանդես է գալիս այդ ենթահարթություններում պատկանող ցանկացած կետի կորդինատների նշանները։ Ակնհայտ է որ ցանկացած կետ որը առաջին քառորդում է ունի դրական կորդինատներ։ Հակառակը նույնպես ճիշտ է այսինքն այն կետը որը ունի դրական կորդինատներ կարող է գտնվել միայն և միայն առաջին քառորդում։ Երկրորդ քառուրդում կետը կգտնվի եթե ունի բացասական $X$ և դրական $Y$ կորդինատ, երորդում եթե և $X$-ը և $Y$-ը բացասական են իսք չորորդում $X>0 \quad Y < 0$: Այս պարզ պնդումը բնականաբար վերաբերվում է նաև $sin$ և $cos$ ֆունկցիաներին։

* $∀ \alpha$ անկյան դեպքում $\alpha + \pi$  անկյան միաոր վեկտորը և $\alpha$ անկյան միաոր վեկտորը ընկած են նույն գծով որը անցնում է $O(0,0)$ սկզբնակետով։ Արդյունքում վեկտորների ծայրակետերը տրամագծորեն հակադիր բեվեռներում կլինեն $⇒$ կունենան հակառակ նշաններ՝ $\sin{(\alpha + \pi)} = -\sin{(\alpha)} \quad \cos{(\alpha + \pi)} = -cos{(\alpha)}$  

```python
fig, ax = create_plot()
dx_orig = 0.9
dy_orig = (1 - dx_orig**2)**(1/2)

create_vector(ax, dx_orig, dy_orig, txt=r'$M_{\alpha}$')
create_vector(ax, -dx_orig, -dy_orig, x_deviation=-0.1, y_deviation=-0.1, txt=r'$M_{\alpha + \pi}$')

add_proection(ax, dx_orig, dy_orig, x_name=r"$X_{\alpha}$", y_name=r"$Y_{\alpha}$",  angle_name=r'$\alpha$')
add_proection(ax, -dx_orig, -dy_orig, x_name=r"$X_{\alpha + \pi}$", y_name=r"$Y_{\alpha + \pi}$",size_angke=0.55,  angle_name_x= -0.25, angle_name_y=0.3,  angle_name=r'$\alpha + \pi$', colr='purple')

plt.title(r'$\alpha$ և $\alpha + \pi $ ՎԵԿՏՈՐՆԵՐԸ ։')
plt.show()
```

Ընդհանրապես $y = f(x)$ կախված ֆունկցիայից ունեն մի շարք հատկություններ ըստ որոնց կարելի է տարանջատել ֆունկցիաները։ Այդպիսի հատկություններից է՝ աճող է թե նվազող, պարբերական է թե ոչ, միարժեք է թե բազմարժեք, աննդհատ է թե ոչ, զույգ է թե կենտ և այլն։ Բոլոր այս հատկությունները հայտնի են դառնում ֆունկցիայի հետազոտման ժամանակ՝ այսինքն իմանալ ֆունկցիայի պահվածքը «վարքը» արգումենտի ամբողջ $R$ բազմությունում։

***
***Սահմանում***

 f(x) ֆունկցիան կոչվում է զույգ, եթե իր արժեքը չի փոխում արգումենտի նշանը փոխելիս, այսինքն ֆունկցիան զույգ է, եթե. $f(x) = f(-x)$ և կենտ, եթե. $f(-x) = -f(x)$


***

զույգ ֆունկցիաների օրինակ է՝ $y = x^2, \quad y= |x|, \quad y = \cos(x) $ կենտ ֆունկցիաներ են՝ $y = x^3, \quad y = kx, \quad y = \sin(x)$ ֆունկցիաները։

Նախ պատկերենք նշված ֆունկցիաների գրաֆիքները հետո քննարկենք $\sin$ և $\cos$ ֆունկցիաների զույգությունը և կենտությունը։  

```python
x = np.linspace(-10, 10, 500)
x_sq = x**2
x_mod = np.abs(x)

x_cub = x**3
x_line = -5*x

plt.figure(figsize=(12, 10)) # Increased figure size for better visibility

# Subplot 1: y = x^2 (Even Function)
ax1 = plt.subplot(221)
ax1.plot(x, x_sq, color='red', label='$y = x^2$')
ax1.set_title('$y = x^2$ (Զույգ ֆունկցիա)') # Armenian for Even Function
ax1.grid(True)
ax1.axhline(0, color='black', linewidth=0.8) # X-axis
ax1.axvline(0, color='black', linewidth=0.8) # Y-axis
ax1.legend()

# Subplot 2: y = |x| (Even Function)
ax2 = plt.subplot(222)
ax2.plot(x, x_mod, color='red', label='$y = |x|$')
ax2.set_title('$y = |x|$ (Զույգ ֆունկցիա)')
ax2.grid(True)
ax2.axhline(0, color='black', linewidth=0.8) # X-axis
ax2.axvline(0, color='black', linewidth=0.8) # Y-axis
ax2.legend()

# Subplot 3: y = x^3 (Odd Function)
ax3 = plt.subplot(223)
ax3.plot(x, x_cub, color='blue', label='$y = x^3$')
ax3.set_title('$y = x^3$ (Կենտ ֆունկցիա)') # Armenian for Odd Function
ax3.grid(True)
ax3.axhline(0, color='black', linewidth=0.8) # X-axis
ax3.axvline(0, color='black', linewidth=0.8) # Y-axis
ax3.legend()

# Subplot 4: y = -5x (Odd Function)
ax4 = plt.subplot(224) # Changed from 223 to 224 to avoid overwriting
ax4.plot(x, x_line, color='blue', label='$y = -5x$')
ax4.set_title('$y = -5x$ (Կենտ ֆունկցիա)')
ax4.grid(True)
ax4.axhline(0, color='black', linewidth=0.8) # X-axis
ax4.axvline(0, color='black', linewidth=0.8) # Y-axis
ax4.legend()

plt.tight_layout() # Adjust layout to prevent overlap
plt.show()
```

Ինչպես երևում է նկարից զույգ ֆունկցիաները նրանք են որոնց գրաֆիքների համար $Y$֊ների առանցքը կարծես լինի «հայելի»։ Եթե $x>0$ բազմության գրաֆիքը պտտենք $Y$ առանցքի շուրջ կստանանք $x<0$֊ի գրաֆիքը՝ այսինքն **զույգ ֆունկցիայի գրաֆիքը սիմետրիկ է $Y$ առանցքի նկատմամբ**։ Նույնը կարելի է ասել կենտ ֆունկցիայի և $X$ առանցքի նկատմամբ։

$ y= \cos(\alpha)$ զույգությունը բխում է այն հանգամանքից որ, որպես $\alpha$ անկյուն վերցրել ենք հենց $X$֊երի առանցքի դրական մասի և ժամ սլաքի ուղղության հակառակ կազմած անկյունը։ Այսպիսով $\alpha$ և $-\alpha$ անկյուն ունեցող վեկտորները սիմետրիկ արտապատկերումներն են $X$ առանցքի նկատմամբ $⇒$ դրանց ծայրակետի $x$ կորդինատները նույնն են իսք $y$ կորդինատները տարբերվում են նշանով $⇒$ $ y= \sin(\alpha)$ կենտ է։

```python
dx_orig = 0.5
dy_orig = (1 - dx_orig**2)**(1/2)

fig, ax = create_plot()

create_vector(ax, dx_orig, dy_orig,  txt=r'$M_\alpha$')
create_vector(ax, dx_orig, -dy_orig, txt=r'$M_{-\alpha}$', )

add_proection(ax, dx_orig, dy_orig,x_name=r"$X_{\alpha}$", y_name=r"$Y_{\alpha}$",  angle_name=r'$+\alpha$')
add_proection(ax, dx_orig, -dy_orig,x_name=r"$X_{-\alpha}$", y_name=r"$Y_{-\alpha}$",  angle_name=r'$-\alpha$', angle_name_x=0.2, angle_name_y= -0.08, colr='purple', pls=False)


plt.title(r'ԱՆԿՅՈՒՆՆԵՐԻ $\alpha$ ԵՎ $ -\alpha $ $\sin$-Ի ԵՎ $\cos$-Ի ԱՐԺԵՔՆԵՐԸ')
plt.show()
```

Նկարում պատկերված է երկու միաոր վեկտորները որոնք կազմում են $α$ և $֊α$ անկյուններ։ Այն որ $\cos{(\alpha)} = \cos{(- \alpha)},  \quad  \sin{(-\alpha)} =  - \sin{(\alpha)}  \quad \forall \alpha$֊ի համար երևում է վերևի նկարից։  

## ՄԻ ՔԱՆԻ ԱՆԿՅՈՒՆՆԵՐԻ $\sin$ և $\cos$ ԱՐԺԵՔՆԵՐԸ։

այժմ տանք մի քանի անկյունների $\sin$ և $\cos$ արժեքները։

երբ $\alpha = 0$ միաոր վեկտորը ընկած է $X$ առանցքի երկայնքով $\Rightarrow$ $\sin{(0)} = 0$, իսք $\cos(0) = 1$: Նույն կերպ երբ $\alpha = \frac{\pi}{2}$ վեկտորը ընկած կլինի $Y$ առանցքի երկայնքով $\Rightarrow$ $\cos{(\frac{\pi}{2})} = 0$, իսք $\sin{(\frac{\pi}{2})} = 1$ $sin{(\pi)} = 0 , \quad cos{(\pi)} = -1, \quad sin{(\frac{3\pi}{2})} = -1, \quad cos{(\frac{3\pi}{2})} = 0 $ ։

```python
dy_orig = 0.5
dx_orig = (1 - dy_orig**2)**(1/2)
fig, ax = create_plot()
create_vector(ax, dx_orig, dy_orig)
add_proection(ax, dx_orig, dy_orig, angle_name=r'$\alpha = \frac{\pi}{6}$' )

ax.plot([dx_orig, dx_orig, 0], [dy_orig, dy_orig-1, 0], color = 'b', linewidth = 1)

ax.text(dx_orig + 0.05, dy_orig-1.05, 'N', fontsize=12, color='red')

plt.title(r"$\frac{\pi}{6}$ ԱՆԿՅԱՆ $\sin$-Ի և $\cos$-Ի ՀԱՇՎԱՐԿԸ։ ")
plt.show()
```

Որպիսի գտնենք $\frac{\pi}{6}$ անկյան սինուսի և կոսինուսի արժեքները դիտարկենք նկարում պատկերված $OMN$ եռանկյունը։ $OM = ON = 1$ շրջանագծի շառավիղն է  $\angle OMN = \frac{pi}{3}$ քանի որ $OMX$ ը ուղղանկյուն եռանկյուն է։ $⇒$ $OMN$-ը հավասարակողմ եռանկյուն է։ իսք $Օx$-ը  հանդիսանում է $NM$ ի միջնուղղահայաց։ նշվածից հետևում է որ $Mx = xN = \sin{(\frac{\pi}{6})} = \frac{1}{2}$: (Այս եզրահանգումը ընթերցողին կարող է ծանոթ լինել դպրոցական երկրաչափությունից նաև այլ կերպ։ ՈՒղղանկյուն եռանկյան $\frac{\pi}{6}$ անկյան դիմաց ընկած էջը հավասար է ներքնաձիքի կեսին։) Օգտվելով եռանկյունաչափական հիմնական նույնությունից կարող ենք հաշվել նաև $cos{(\frac{\pi}{6})} = \sqrt{1 - \frac{1}{4}} = \frac{\sqrt{3}}{2}$: Ընթերցողին առաջարկում ենք հաշվել  $\frac{5 \pi}{6}, \frac{7 \pi}{6}, \frac{11 \pi}{6}$ անկյունների $\sin$ և $\cos$ արժեքները։

Այժմ գտնենք $\frac{\pi}{4}$ անկյան $\sin$-ը և $\cos$-ը։ Սկզբից պատկերենք միաոր վեկտոր որը $X$-էրի հետ կազմում է $\frac{\pi}{4}$ անկյուն։

```python
dx_orig = (2**(1/2))/2
dy_orig = (2**(1/2))/2
fig, ax = create_plot()
create_vector(ax, dx_orig, dy_orig)
add_proection(ax, dx_orig, dy_orig, angle_name=r'$\alpha = \frac{\pi}{4}$' )

ax.plot([dx_orig, dx_orig], [0, dy_orig], color='b', linestyle='-', linewidth=1) # Vertical dashed line to x-axis
ax.plot([0, dx_orig], [dy_orig, dy_orig], color='b', linestyle='-', linewidth=1) # Horizontal dashed line to y-axis

plt.title(r"$\frac{\pi}{4}$ ԱՆԿՅԱՆ $\sin$-Ի և $\cos$-Ի ՀԱՇՎԱՐԿԸ։ ")
plt.show()
```

$OYMX$ քառանկյան բոլոր անկյունները $\frac{\pi}{2}$ է, իսք անկյունագիծը կողմերի հետ կազմում է $\frac{\pi}{4}$ անկյուն $⇒$ այն քառակուսի է։ Նշանակուն է $x = y$ այսինքն $\sin{(\frac{\pi}{4})} = \cos{(\frac{\pi}{4})} $: Օգտվելով հիմնական նույնությունից կստանանք $\sin{(\frac{\pi}{4})} = \cos{(\frac{\pi}{4})} = \frac{\sqrt{2}}{2}$: Դժվար չէ գտնել $\frac{\pi}{4} + n \frac{\pi}{2}, \quad n \in N $ անկյունների $\sin$-ը և $\cos$-ը։

$\frac{\pi}{3} $ անկյան $\sin$ և $\cos$ ֆունկցիաների արժեքների հաշվարկը կթողնենք ընթերցողին։

Այսպիսով ստացված արդյունքները կարող ենք գրել աղյուսակի տեսքով՝

## $\sin$ և $\cos$ ՖՈՒՆԿՑԻԱՆԵՐԻ ԳՐԱՖԻԿՆԵՐԸ։ ՁԱՅՆԸ ՈՐՊԵՍ ՆԵՐԴԱՇՆԱԿ ՖՈՒՆԿՑԻԱ։

Այժմ ստանանք դիտարկված ֆունկցիաների գրաֆիկը։ Որպես անկյան արժեքներ վերցնենք $[-2\pi, 2\pi]$ միջակայքը։


```python
alpha = np.linspace(-2*np.pi, 2*np.pi, 720)
sin_alp = np.sin(alpha)

first_quadrant_famus = np.array([0, np.pi/6, np.pi/4, np.pi/3, np.pi/2])

all_positive_famous = np.array([])
for q_mult in range(4):
    all_positive_famous = np.concatenate((all_positive_famous, first_quadrant_famus + q_mult * np.pi/2))

all_positive_famous = np.unique(all_positive_famous)
all_positive_famous = all_positive_famous[all_positive_famous <= 2*np.pi]
if not np.isclose(all_positive_famous[-1], 2*np.pi):
    all_positive_famous = np.append(all_positive_famous, 2*np.pi)

famous_alpha = np.concatenate((-all_positive_famous, all_positive_famous))
famous_alpha = np.unique(famous_alpha)
famous_alpha.sort()

sin_famous = np.sin(famous_alpha)

# Define famous angle labels mapping numerical values to LaTeX strings
pi_labels_map = {
    0.0: r'$0$',
    np.pi / 6: r'$\frac{\pi}{6}$',
    np.pi / 4: r'$\frac{\pi}{4}$',
    np.pi / 3: r'$\frac{\pi}{3}$',
    np.pi / 2: r'$\frac{\pi}{2}$',
    2 * np.pi / 3: r'$\frac{2\pi}{3}$',
    3 * np.pi / 4: r'$\frac{3\pi}{4}$',
    5 * np.pi / 6: r'$\frac{5\pi}{6}$',
    np.pi: r'$\pi$',
    7 * np.pi / 6: r'$\frac{7\pi}{6}$',
    5 * np.pi / 4: r'$\frac{5\pi}{4}$',
    4 * np.pi / 3: r'$\frac{4\pi}{3}$',
    3 * np.pi / 2: r'$\frac{3\pi}{2}$',
    5 * np.pi / 3: r'$\frac{5\pi}{3}$',
    7 * np.pi / 4: r'$\frac{7\pi}{4}$',
    11 * np.pi / 6: r'$\frac{11\pi}{6}$',
    2 * np.pi: r'$2\pi$',
}

# Create combined list of tick positions and labels for both positive and negative famous_alpha values
x_ticks_positions = []
x_ticks_labels = []

# Sort famous_alpha to ensure ticks are in order
sorted_famous_alpha = np.sort(famous_alpha)

for val in sorted_famous_alpha:
    label_found = False
    for pi_val, label_str in pi_labels_map.items():
        if np.isclose(val, pi_val):
            x_ticks_positions.append(val)
            x_ticks_labels.append(label_str)
            label_found = True
            break
        elif np.isclose(val, -pi_val) and pi_val != 0:
            x_ticks_positions.append(val)
            # For negative values, prepend '-' directly after the opening '$' for other labels
            x_ticks_labels.append(label_str[0] + '-' + label_str[1:])
            label_found = True
            break

# Define simplified y-axis tick values and labels for clarity
y_ticks_values = np.array([-1, -0.5, 0, 0.5, 1])
y_ticks_labels = [
    r'$-1$',
    r'$-\frac{1}{2}$',
    r'$0$',
    r'$\frac{1}{2}$',
    r'$1$'
]

plt.figure(figsize=(15, 10)) # Increase figure size for better readability of labels

# Subplot 1: sin function
ax1 = plt.subplot(211)
ax1.plot(famous_alpha, sin_famous, color='tab:blue', marker='o', linestyle='None', label='ՀԱՅՏՆԻ ԱՆԿՅՈՒՆՆԵՐԸ')
ax1.plot(alpha, sin_alp, color='black', label=r'$\sin(\alpha)$ ԳՐԱՖԻԿԸ') # Corrected LaTeX in label
ax1.set_title(r'$\sin(\alpha)$ ՖՈՒՆԿՑԻԱԻ ԳՐԱՖԻԿԸ') # Corrected LaTeX in title
ax1.grid(True, linestyle='--', alpha=0.7)
ax1.axhline(0, color='black', linewidth=1.0) # X-axis
ax1.axvline(0, color='black', linewidth=1.0) # Y-axis
ax1.set_xticks(x_ticks_positions)
ax1.set_xticklabels(x_ticks_labels, rotation=45, ha='right', fontsize=10) # Adjust font size and rotation
ax1.set_yticks(y_ticks_values)
ax1.set_yticklabels(y_ticks_labels, fontsize=10)
ax1.set_xlabel('ԱՆԿՅՈՒՆԸ (ՌԱԴԻԱՆ)', fontsize=12)
ax1.set_ylabel(r'$\sin(\alpha)$', fontsize=12) # Corrected LaTeX in ylabel
ax1.legend()


# Subplot 2: cos function
ax2 = plt.subplot(212)
ax2.plot(famous_alpha, np.cos(famous_alpha), color='tab:blue', marker='o', linestyle='None', label='ՀԱՅՏՆԻ ԱՆԿՅՈՒՆՆԵՐԸ')
ax2.plot(alpha, np.cos(alpha), color='black', label=r'$\cos(\alpha)$ ԳՐԱՖԻԿԸ') # Corrected LaTeX in label
ax2.set_title(r'$\cos(\alpha)$ ՖՈՒՆԿՑԻԱԻ ԳՐԱՖԻԿԸ') # Corrected LaTeX in title
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.axhline(0, color='black', linewidth=1.0) # X-axis
ax2.axvline(0, color='black', linewidth=1.0) # Y-axis
ax2.set_xticks(x_ticks_positions)
ax2.set_xticklabels(x_ticks_labels, rotation=45, ha='right', fontsize=10) # Adjust font size and rotation
ax2.set_yticks(y_ticks_values)
ax2.set_yticklabels(y_ticks_labels, fontsize=10)
ax2.set_xlabel('ԱՆԿՅՈՒՆԸ (ՌԱԴԻԱՆ)', fontsize=12)
ax2.set_ylabel(r'$\cos(\alpha)$', fontsize=12) # Corrected LaTeX in ylabel
ax2.legend()

plt.tight_layout()
plt.show()
```

Որպես գործնական օրինակ դիտարկենք ձայնը որի հետ մենք բոլորս ամեն օր առնչվում ենք։ Ինչպես գիտենք ձայնը օդի տատանում է որը կարող է առաջանալ տարբեր եղանակներով (մարդու ձայնալարերի տատանումները, բարձրախոսի մեմբրանի տատանումները, ․․․)։ Մարդու ականջները կարող են լսել միջինում 20 ից 20․000 հց հաճախությամբ տատանումներ «կան նաև բացառություններ»։ Մի փոքր այն մասին թե ինչ է  հաճախականությունը՝ դա միաոր ժամանակում կատարած լրիվ տատանումների թիվն է։ Մեկ տատանումը դա այն միջակայքն է երբ մեծությունը սկսում է կրկնել իր նախքինում ունեցած արժեքները։ Այժմ դիտարկենք $y = A_0\sin{(ωt)}$, $t$ ժամանակից կախված ֆունկցիան։ $\omega$֊ն կոչվում է անկյունային հաճախություն ակնհայտ է որ որքան մեծացնենք $\omega$֊ն այդքան շատ կլինի միաոր ժամանակում կատարած տատանումների թիվը, $A_0$֊ն կոչվում է ամպլիտուդ այստեղ էլ ակնհայտ է որ $y \in [-A_0, A_0 ]$։ Ասվածի պարզեցման համար պատկերենք երկու ուսանելի տատանումներ որից առաջինը ժամանակի ընթացքում արագանում է։ Իսք երկրորդը մարում է։ Այլ կերպ ասած առաջինում $\omega$ ֊ն աճող ֆունկցիա է ժամանակից $\omega(t) = \omega_0 + kt \quad k>0$ Իսք երկրորդի ամպլիտուդն է նվազող ֆունկցիա ժամանակից։      

```python

sampling_rate = 1000
duration = 5
t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)


omega_0_1 = 2 * np.pi * 1
k_freq = 2 * np.pi * 0.5
amplitude_1 = 1.0

accelerating_oscillation = amplitude_1 * np.sin(omega_0_1 * t + 0.5 * k_freq * t**2)


frequency_2 = 2.0
amplitude_0_2 = 1.0
k_amp = 0.5

amplitude_over_time = amplitude_0_2 * np.exp(-k_amp * t)

damping_oscillation = amplitude_over_time * np.sin(2 * np.pi * frequency_2 * t)


plt.figure(figsize=(12, 10))

plt.subplot(2, 1, 1)
plt.plot(t, accelerating_oscillation, color='blue')
plt.title(r'ԱՃՈՂ ՀԱՃԱԽՈՒԹՅԱՄԲ ՏԱՏԱՆՈՒՄ ($\omega(t) = \omega_0 + kt$)')
plt.xlabel('ԺԱՄԱՆԱԿԸ (Վ)')
plt.ylabel('ԱՄՊԼԻՏՈՒԴԸ')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(t, damping_oscillation, color='red')
plt.title(r'ՄԱՐՈՂ ՏԱՏԱՆՈՒՄ ($A(t) = A_0 e^{-kt}$)')
plt.xlabel('ԺԱՄԱՆԱԿԸ (Վ)')
plt.ylabel('ԱՄՊԼԻՏՈՒԴԸ')
plt.grid(True)

plt.tight_layout()
plt.show()
```

Չշեղվելով թեմայից ասենք որ երաժշտական նոտաները դրանք սինուսոիդներ (ժամանակից կախված սինուս ֆունկցիա) էն տարբեր հիմնական հաճախություններով և հարմոնիկներից ( հարմոնիկների մասին կարող եք գտնել տեղեկություն համացանցից )։ այժմ ստանանք հինգերորդ օկտավաի դո նոտան պատկերենք այն 10 մվ ժամանակով հետո լսենք։

```python

frequency_hz = 523.25
sampling_rate = 44100
duration = 2.0

t_audio = np.linspace(0., duration, int(sampling_rate * duration), endpoint=False)

samples =  np.sin(2. * np.pi * frequency_hz * t_audio)

plt.figure(figsize=(10, 4))

plot_duration_ms = 10
plot_samples = int(sampling_rate * (plot_duration_ms / 1000.0))

plot_samples = min(plot_samples, len(samples))

t_plot = np.linspace(0., plot_duration_ms / 1000.0, plot_samples, endpoint=False)

plt.plot(t_plot, samples[:plot_samples], color='blue')
plt.title(f'ԴՈ ՆՈՏԱԻ ԱԼԻՔԻ ՁևԸ ({frequency_hz} Hz) - ԱՌԱՋԻՆ {plot_duration_ms} մվ')
plt.xlabel('ԺԱՄԱՆԱԿԸ (Վ)')
plt.ylabel('ԱՄՊԼԻՏՈՒԴԸ')
plt.grid(True)
plt.show()

print(f"Generating C5 note at {frequency_hz} Hz for {duration} seconds...")
display(Audio(samples, rate=sampling_rate))
print("Playback initiated.")
```

$X_{α + β}$ ֊ն ${\alpha+\beta}$ անկյան $\cos$֊ն է որը՝ հավասար $OR$ ֊ից հանած $SQ$ ֊ի երկարությունը։  $OR$֊ը իր հերթին $OQ$ ֊ի պրոեկցիան է $X$ առանցքի վրա $OR = OQ \cos{\alpha}$: Այժմ ուշադիր նայենք նկարին կտեսնենք որ $OQ$ ֊ն  $\overrightarrow{OM}_{\alpha+\beta}$ վեկտորի $X$ պրոեկցիան է $α$ անկյամբ պտտված կորդինատային  համակարգում  $⇒ \quad OQ = \cos{\beta}$: Ստացվում է $\cos{(\alpha + \beta)} = \cos{(\alpha)}\cos{(\beta)} - SQ $: $SQ$֊ն հաշվելու համար ցույց տանք որ $∠ SM_{\alpha + \beta}Q = \alpha $։  Իսկապես նշված անկյունը  $∠ OM_{\alpha + \beta}Q = \frac{\pi}{2} - \beta$ և $∠ OM_{\alpha + \beta}S = \frac{\pi}{2} - \alpha - \beta$ անկյունների տարբերությունն է որը $\alpha$ է։ Ակնհայտ է որ $M_{\alpha + \beta}Q = \sin{(\beta)}$ իսք $SQ = M_{\alpha + \beta} Q \sin{(\alpha)} \quad ⇒ SQ = \sin{(\alpha)} \sin{(\beta)}$ ստացվեց՝

$$
\boldsymbol{\cos{(\alpha + \beta)} =  \cos{(\alpha)}\cos{(\beta)} - \sin{(\alpha)} \sin{(\beta)} }
$$
Ակնհայտ է որ եթե $\alpha = \beta$ ապա՝

$$
\boldsymbol{\cos{( 2 \alpha)} =  \cos^2{(\alpha)} - \sin^2{(\alpha)}}
$$


$Y_{α + β}$ ֊ն ${\alpha+\beta}$ անկյան $\sin$֊ն է որը՝ հավասար $RQ$ ֊ի և $SM_{\alpha + \beta}$ ֊ի գումարին։  $RQ$֊ն իր հերթին $OQ$ ֊ի պրոեկցիան է $Y$ առանցքի վրա $RQ = OQ \sin{\alpha}$: ինչպես ցույց տվեցինք վերևում  $ OQ = \cos{\beta}$: Ստացվում է $\sin{(\alpha + \beta)} = \sin{(\alpha)}\cos{(\beta)} + SM_{\alpha + \beta} $:

$△SM_{\alpha + \beta} Q$ ից   որտեղ  $∠ SM_{\alpha + \beta}Q =  \alpha$ և  $M_{\alpha + \beta}Q = \sin{(\beta)}$ կարող ենք գրել՝  $SM_{\alpha + \beta} = \sin{(\beta)} \cos{(\alpha)} $ ստացվեց՝

$$
\boldsymbol{\sin{(\alpha + \beta)} =  \sin{(\alpha)}\cos{(\beta)} + \cos{(\alpha)} \sin{(\beta)} }
$$

Ակնհայտ է որ եթե $\alpha = \beta$ ապա՝

$$
\boldsymbol{ \sin{( 2 \alpha) } =  2\sin{(\alpha)}\cos{(\alpha)} }
$$


Անկյունների տարբերության եռանկյունաչափական ֆունկցիայի արժեքները կարելի է ստանալ նման ձևով որը շատ նման է վերևում բերված ապացույցներին։ Այստեղ մենք կոգտվենք $\alpha - \beta = \alpha + (- \beta)$ և այն փաստերից որ $\sin{(-\alpha)} = - \sin{(\alpha)}$ և $\cos{(-\alpha)} = \cos{(\alpha)}$ կստացվի՝

$$
\boldsymbol{\cos{(\alpha - \beta)} =  \cos{(\alpha)}\cos{(\beta)} + \sin{(\alpha)} \sin{(\beta)} }
$$


$$
\boldsymbol{\sin{(\alpha - \beta)} =  \sin{(\alpha)}\cos{(\beta)} - \cos{(\alpha)} \sin{(\beta)} }
$$




### ԵՌԱՆԿՅՈՒՆԱՉԱՓԱԿԱՆ ՖՈՒՆԿՑԻԱՆԵՐԻ ԳՈՒՄԱՐԸ և ՏԱՐԲԵՐՈՒԹՅՈՒՆԸ։

Դիցուկ ունենք որևէ $\alpha$ և $\beta$ անկյուններ վերևում ցույց տվել ենք որ $\sin{(\alpha \pm \beta)} =  \sin{(\alpha)}\cos{(\beta)} \pm \cos{(\alpha)} \sin{(\beta)}$։ Այստեղ գրված երկու արտահայտությունները գումարենք իրար կստանանք՝ $\sin{(\alpha + \beta)} + \sin{(\alpha - \beta)} = 2 \sin{(\alpha)}\cos{(\beta)}$

### ԵՌԱՆԿՅՈՒՆԱՉԱՓԱԿԱՆ ՖՈՒՆԿՑԻԱՆԵՐԻ ԳՈՒՄԱՐԸ և ՏԱՐԲԵՐՈՒԹՅՈՒՆԸ։

Դիցուկ ունենք որևէ $x$ և $y$ անկյուններ վերևում ցույց տվել ենք որ՝

1.  $\sin{(x + y)} =  \sin{(x)}\cos{(y)} + \cos{(x)} \sin{(y)}$
2.  $\sin{(x - y)} =  \sin{(x)}\cos{(y)} - \cos{(x)} \sin{(y)}$
3.  $\cos{(x + y)} =  \cos{(x)}\cos{(y)} - \sin{(x)} \sin{(y)}$
4.  $\cos{(x - y)} =  \cos{(x)}\cos{(y)} + \sin{(x)} \sin{(y)}$

Կատարենք մի քանի պարզ գործողություններ։ Առաջին հավասարումը գումարենք երկրորդին՝

$$ \boldsymbol\sin{(x + y)} + \sin{(x - y)} = 2 \sin{(x)}\cos{(y)} $$

Առաջինից հանենք երկրորդը՝

$$ \boldsymbol{\sin{(x + y)} - \sin{(x - y)} = 2 \cos{(x)}\sin{(y)}} $$

Գումարենք երրորդ և չորորդ հավասարումները՝

$$ \boldsymbol{\cos{(x + y)} + \cos{(x - y)} = 2 \cos{(x)}\cos{(y)}} $$

Երրորդից հանենք չորորդը՝
$$ \boldsymbol{\cos{(x + y)} - \cos{(x - y)} = -2 \sin{(x)} \sin{(y)}} $$

Նշանակենք՝ $x + y = \alpha$ և $x-y = \beta$: Ակնհայտ է որ $\forall \alpha, \beta$֊ի համար կգտնվեն այնպիսի $x = \frac{\alpha + \beta}{2}$ և  $y = \frac{\alpha-\beta}{2}$ որ տեղի ունենա նշանակումը։

Ստացված հավասարումներում $x, y, x+y, x-y$֊ը փոխարինենք նշանակումներով կստանանք։

Սինուսների գումարի բանաձևը.
$$ \boldsymbol{\sin{\alpha} + \sin{\beta} = 2 \sin{\left(\frac{\alpha+\beta}{2}\right)} \cos{\left(\frac{\alpha-\beta}{2}\right)}} $$

Սինուսների տարբերության բանաձևը.
$$ \boldsymbol{\sin{\alpha} - \sin{\beta} = 2 \cos{\left(\frac{\alpha+\beta}{2}\right)} \sin{\left(\frac{\alpha-\beta}{2}\right)}} $$

Կոսինուսների գումարի բանաձևը.
$$ \boldsymbol{\cos{\alpha} + \cos{\beta} = 2 \cos{\left(\frac{\alpha+\beta}{2}\right)} \cos{\left(\frac{\alpha-\beta}{2}\right)}} $$

Կոսինուսների տարբերության բանաձևը.
$$ \boldsymbol{\cos{\alpha} - \cos{\beta} = -2 \sin{\left(\frac{\alpha+\beta}{2}\right)} \sin{\left(\frac{\alpha-\beta}{2}\right)}} $$

### Արտադրյալից գումարի կամ տարբերության բանաձևեր

Այժմ դուրս բերենք արտադրյալի բանաձևերը՝ օգտագործելով արդեն ստացված գումարի և տարբերության բանաձևերը։

Մենք ունենք հետևյալ բանաձևերը.
1.  $\sin{(\alpha + \beta)} =  \sin{(\alpha)}\cos{(\beta)} + \cos{(\alpha)} \sin{(\beta)}$
2.  $\sin{(\alpha - \beta)} =  \sin{(\alpha)}\cos{(\beta)} - \cos{(\alpha)} \sin{(\beta)}$
3.  $\cos{(\alpha + \beta)} =  \cos{(\alpha)}\cos{(\beta)} - \sin{(\alpha)} \sin{(\beta)}$
4.  $\cos{(\alpha - \beta)} =  \cos{(\alpha)}\cos{(\beta)} + \sin{(\alpha)} \sin{(\beta)}$

**1. $\sin{(\alpha)}\cos{(\beta)}$ արտադրյալի դուրսբերում**

Գումարենք (1) և (2) բանաձևերը.

$(\sin{(\alpha + \beta)}) + (\sin{(\alpha - \beta)}) = (\sin{(\alpha)}\cos{(\beta)} + \cos{(\alpha)} \sin{(\beta)}) + (\sin{(\alpha)}\cos{(\beta)} - \cos{(\alpha)} \sin{(\beta)})$

Միևնույն անդամները կկրճատվեն, և կստանանք՝

$\sin{(\alpha + \beta)} + \sin{(\alpha - \beta)} = 2\sin{(\alpha)}\cos{(\beta)}$

Այստեղից՝
$$ \boldsymbol{\sin{(\alpha)}\cos{(\beta)} = \frac{1}{2} [\sin{(\alpha + \beta)} + \sin{(\alpha - \beta)}]} $$

**2. $\cos{(\alpha)}\cos{(\beta)}$ արտադրյալի դուրսբերում**

Գումարենք (3) և (4) բանաձևերը.

$(\cos{(\alpha + \beta)}) + (\cos{(\alpha - \beta)}) = (\cos{(\alpha)}\cos{(\beta)} - \sin{(\alpha)} \sin{(\beta)}) + (\cos{(\alpha)}\cos{(\beta)} + \sin{(\alpha)} \sin{(\beta)})$

Միևնույն անդամները կկրճատվեն, և կստանանք՝

$\cos{(\alpha + \beta)} + \cos{(\alpha - \beta)} = 2\cos{(\alpha)}\cos{(\beta)}$

Այստեղից՝
$$ \boldsymbol{\cos{(\alpha)}\cos{(\beta)} = \frac{1}{2} [\cos{(\alpha + \beta)} + \cos{(\alpha - \beta)}]} $$

**3. $\sin{(\alpha)}\sin{(\beta)}$ արտադրյալի դուրսբերում**

Հանենք (3) բանաձևը (4) բանաձևից.

$(\cos{(\alpha - \beta)}) - (\cos{(\alpha + \beta)}) = (\cos{(\alpha)}\cos{(\beta)} + \sin{(\alpha)} \sin{(\beta)}) - (\cos{(\alpha)}\cos{(\beta)} - \sin{(\alpha)} \sin{(\beta)})$

Փակագծերը բացելուց հետո՝

$\cos{(\alpha - \beta)} - \cos{(\alpha + \beta)} = \cos{(\alpha)}\cos{(\beta)} + \sin{(\alpha)} \sin{(\beta)} - \cos{(\alpha)}\cos{(\beta)} + \sin{(\alpha)} \sin{(\beta)}$

Միևնույն անդամները կկրճատվեն, և կստանանք՝

$\cos{(\alpha - \beta)} - \cos{(\alpha + \beta)} = 2\sin{(\alpha)}\sin{(\beta)}$

Այստեղից՝
$$ \boldsymbol{\sin{(\alpha)}\sin{(\beta)} = \frac{1}{2} [\cos{(\alpha - \beta)} - \cos{(\alpha + \beta)}]} $$
